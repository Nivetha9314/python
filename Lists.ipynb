```python
# Generate 20 random numbers (between 1 and 10) and add it to a list
# Display the first and the last number
# Sort the numbers and display the list
# Print the mode (number appearing maximum number of times)
```

```python
import random

n = []

for i in range(20):
    num = random.randint(1, 10)
    n.append(num)

print(n)

# Display the first and the last number
print("first number", n[0])
print("last number", n[-1])

# Sort the numbers and display the list
n.sort()
print(n)

# Print the mode (number appearing maximum number of times)
```

```python
numbers = [10, 15, 22, 7, 8, 13, 20]

def get_even_numbers(numbers):
    n=[]
    for i in numbers:
        if i%2==0:
            n.append(i)
    return n

get_even_numbers(numbers)
```
