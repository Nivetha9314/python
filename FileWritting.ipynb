```python
## creating and writting to file

file='Example.txt'
with open(file,'w') as f:
    f.write("This is line A")
```

```python
## Read file

with open(file,'r') as testwrittingfile:
    print(testwrittingfile.read())
```

```python
## write lines to file -multiple line

with open(file,'w') as writtingfile:
    writtingfile.write("This is line B")
    writtingfile.write("This is line C")
```

```python
## sample list of text
Lines=["Hello\n", "How are you\n","Welcome to python programming\n"]

print(Lines)

for i in Lines:
    print(i)
```

```python
## write the strings in the list to text file

with open(file,'w') as writefile:
    for line in Lines:
        print(line)
        writefile.write(line)
```

## Copy a file

```python
## copy file to another

with open('Example.txt','r') as readfile:
    with open('Example1.txt','w') as writefile:
        for line in readfile:
            writefile.write(line)
```

```python
import csv

rows = []

with open("test.csv", "r") as f:
    data_reader = csv.reader(f, delimiter=",")

    for line in data_reader:
        if line[0] == "Ramya":
            line[1] = "99.5"

        rows.append(line)

with open("test.csv", "w", newline="") as f:
    writer = csv.writer(f, delimiter=",")
    writer.writerows(rows)
```
