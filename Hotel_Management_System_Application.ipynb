## Room Class

```python
from abc import ABC, abstractmethod
class rooms(ABC):
    def __init__(self,room_no,room_type,price_per_day,availablity_status):
        self.room_no= room_no
        self.room_type=room_type
        self.price_per_day=price_per_day
        self.__availablity_status=availablity_status

    @abstractmethod
    def roomtype(self):
        pass

    def getstatus(self):
        return self.__availablity_status

    def setstatus(self, status):
      if status in ["Available", "Booked"]:
         self.__availablity_status = status
      else:
         print("Invalid status")

    def display(self):
       print(f'Room no: {self.room_no}, Room Type: {self.room_type}, '
          f'Price per day: {self.price_per_day}, '
          f'Availability status: {self.getstatus()}')
```

## Room Types

```python
class singleroom(rooms):
    def __init__(self, room_no, price_per_day, availability_status):
        super().__init__(room_no, "Single", price_per_day, availability_status)
    def roomtype(self):
        print("Single room - Basic facilities")

class doubleroom(rooms):
    def __init__(self, room_no, price_per_day, availability_status):
        super().__init__(room_no, "Double", price_per_day, availability_status)
    def roomtype(self):
        print("Double room - Extra bed facility")

class deluxroom(rooms):
    def __init__(self, room_no, price_per_day, availability_status):
        super().__init__(room_no, "Delux", price_per_day, availability_status)
    def roomtype(self):
        print("Delux room - Breakfast and Wi-Fi")
```

## Customer Class

```python
## customer class

class customer:
    def __init__(self,customer_id,customer_name,phone_no,email):
        self.customer_id=customer_id
        self.customer_name = customer_name
        self.phone_no=phone_no
        self.email=email

    def display(self):
        print(f'customer id: {self.customer_id}, customer name:{self.customer_name}, phone no: {self.phone_no}, email: {self.email}')
    
```

## Booking Class

```python
class Booking:
    def __init__(self, booking_id, customer, room, check_in_date, number_of_days):
        self.booking_id = booking_id
        self.customer = customer
        self.room = room
        self.check_in_date = check_in_date
        self.number_of_days = number_of_days

    def display(self):
        print(
            f"Booking id: {self.booking_id}, "
            f"Customer: {self.customer.customer_name}, "
            f"Room: {self.room.room_no}, "
            f"Check-in date: {self.check_in_date}, "
            f"Number of days: {self.number_of_days}"
        )
```

## Checkout Class

```python
class Checkout:
    def __init__(self, booking):
        self.booking = booking

    def calculate_room_cost(self):
        return self.booking.room.price_per_day * self.booking.number_of_days

    def add_charge(self, additional_charge):
        room_cost = self.calculate_room_cost()
        total = room_cost + additional_charge
        return total

    def display_bill(self, additional_charge):
        room_cost = self.calculate_room_cost()
        total = room_cost + additional_charge

        print("----- Final Bill -----")
        print("Customer:", self.booking.customer.customer_name)
        print("Room No:", self.booking.room.room_no)
        print("Number of Days:", self.booking.number_of_days)
        print("Room Cost:", room_cost)
        print("Additional Charge:", additional_charge)
        print("Total Bill:", total)

    def checkout_room(self):
        self.booking.room.setstatus("Available")
        print("Room is now available")
```

```python
# Existing rooms

rooms_list = []

rooms_list.append(s1)
rooms_list.append(s2)
rooms_list.append(d1)
rooms_list.append(dl1)


# Customer list
customer_list = []


# Booking list
booking_list = []


# Hotel Management Menu
while True:

    print("\n----- HOTEL MANAGEMENT SYSTEM -----")
    print("1. Add Room")
    print("2. View All Rooms")
    print("3. Search Room")
    print("4. Add Customer")
    print("5. Book Room")
    print("6. View Bookings")
    print("7. Checkout")
    print("8. Generate Bill")
    print("9. Cancel Booking")
    print("10. Exit")

    choice = int(input("Enter your choice: "))


    # 1. Add Room
    if choice == 1:

        room_no = int(input("Enter room number: "))

        print("1. Single")
        print("2. Double")
        print("3. Delux")

        room_type = int(input("Enter room type: "))
        price = int(input("Enter price per day: "))

        if room_type == 1:
            new_room = singleroom(
                room_no,
                price,
                "Available"
            )

        elif room_type == 2:
            new_room = doubleroom(
                room_no,
                price,
                "Available"
            )

        elif room_type == 3:
            new_room = deluxroom(
                room_no,
                price,
                "Available"
            )

        else:
            print("Invalid room type")
            continue

        rooms_list.append(new_room)

        print("Room added successfully")


    # 2. View All Rooms
    elif choice == 2:

        print("\n----- ALL ROOMS -----")

        for i in rooms_list:
            i.display()


    # 3. Search Room
    elif choice == 3:

        room_no = int(input("Enter room number: "))

        for i in rooms_list:

            if i.room_no == room_no:
                i.display()
                break

        else:
            print("Room not found")


    # 4. Add Customer
    elif choice == 4:

        customer_id = int(input("Enter customer id: "))
        customer_name = input("Enter customer name: ")
        phone_no = input("Enter phone number: ")
        email = input("Enter email: ")

        new_customer = customer(
            customer_id,
            customer_name,
            phone_no,
            email
        )

        customer_list.append(new_customer)

        print("Customer added successfully")


    # 5. Book Room
    elif choice == 5:

        customer_id = int(input("Enter customer id: "))

        selected_customer = None

        for c in customer_list:

            if c.customer_id == customer_id:
                selected_customer = c
                break

        if selected_customer is None:
            print("Customer not found")
            continue


        room_no = int(input("Enter room number: "))

        selected_room = None

        for r in rooms_list:

            if r.room_no == room_no and r.getstatus() == "Available":
                selected_room = r
                break

        if selected_room is None:
            print("Room not available")
            continue


        booking_id = int(input("Enter booking id: "))
        check_in_date = input("Enter check-in date: ")
        number_of_days = int(input("Enter number of days: "))


        selected_room.setstatus("Booked")


        new_booking = Booking(
            booking_id,
            selected_customer,
            selected_room,
            check_in_date,
            number_of_days
        )


        booking_list.append(new_booking)

        print("Room booked successfully")


    # 6. View Bookings
    elif choice == 6:

        print("\n----- ALL BOOKINGS -----")

        if len(booking_list) == 0:
            print("No bookings available")

        else:

            for i in booking_list:
                i.display()


    # 7. Checkout
    elif choice == 7:

        booking_id = int(input("Enter booking id: "))

        selected_booking = None

        for b in booking_list:

            if b.booking_id == booking_id:
                selected_booking = b
                break


        if selected_booking is None:
            print("Booking not found")
            continue


        checkout = Checkout(selected_booking)

        checkout.checkout_room()

        print("Checkout completed successfully")


    # 8. Generate Bill
    elif choice == 8:

        booking_id = int(input("Enter booking id: "))

        selected_booking = None

        for b in booking_list:

            if b.booking_id == booking_id:
                selected_booking = b
                break


        if selected_booking is None:
            print("Booking not found")
            continue


        additional_charge = int(
            input("Enter additional charges: ")
        )


        checkout = Checkout(selected_booking)

        checkout.display_bill(additional_charge)


    # 9. Cancel Booking
    elif choice == 9:

        booking_id = int(input("Enter booking id: "))

        selected_booking = None

        for b in booking_list:

            if b.booking_id == booking_id:
                selected_booking = b
                break


        if selected_booking is None:
            print("Booking not found")
            continue


        selected_booking.room.setstatus("Available")

        booking_list.remove(selected_booking)

        print("Booking cancelled successfully")


    # 10. Exit
    elif choice == 10:

        print("Thank you!")

        break


    # Invalid choice
    else:

        print("Invalid choice")
```
