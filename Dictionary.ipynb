```python
school = {
    "Grade 10": [
        {
            "roll_no": 1,
            "name": "Arun",
            "age": 15,
            "subjects": {"Math", "Science", "English"}
        },
        {
            "roll_no": 2,
            "name": "Priya",
            "age": 15,
            "subjects": {"Math", "Tamil", "Social"}
        },
        {
            "roll_no":3,
            "name":"sneha",
            "age":15,
            "subjects":{"Hindi","Evs","GK"}
        }
    ],
    "Grade 11": [
        {
            "roll_no": 1,
            "name": "Ravi",
            "age": 16,
            "subjects": {"Physics", "Chemistry", "Math"}
        },
        {
            "roll_no": 2,
            "name": "kumar",
            "age": 16,
            "subjects": {"commerce", "economics", "Math"}
        },
        {
            "roll_no": 3,
            "name": "meera",
            "age": 16,
            "subjects": {"cs", "Chemistry", "Math"}
        }
    ]
}
```

```python
def add_student(grade,roll_no,name,age,subjects):
  student={
    "roll_no":roll_no,
    "name":name,
    "age":age,
    "subjects":set(subjects)
  }
  school[grade].append(student)

add_student(
    "Grade 10",
    4,
    "Karthiga",
    15,
    ["Maths","science","tamil"]
)

print(school["Grade 10"])
```

```python
def add_subject(grade,roll_no,subject):
    for student in school[grade]:
        if student["roll_no"]==roll_no:
            student["subjects"].add(subject)
            print("subject added sucessfully")
            return
        
```

```python
add_subject("Grade 10", 1,"Tamil")
print(school["Grade 10"])
```

```python
def get_student_name(grade):
    for student in school[grade]:
        
        print(student["name"])

get_student_name("Grade 11")            
```

```python
##Write a function to display all unique subjects offered across the entire school.

def get_subjects():
    subject=set()
    for grade in school:
        
      for student in school[grade]:
        subject.update(student["subjects"])
    print(subject)

get_subjects()
        
```

```python
##Try to print one student’s information in a formatted way using tuples.

def print_student(grade,roll_no):
    for student in school[grade]:
        if student["roll_no"]==roll_no:
            student_info=(
                student["roll_no"],
                student["name"],
                student["age"],
                student["subjects"]
                    
            )
            print("roll_no:", student_info[0])
            print("name:", student_info[1])
            print("age:", student_info[2])
            print("subjects:", student_info[3])
print_student("Grade 10",1)
```
