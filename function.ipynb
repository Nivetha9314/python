```python
## positive or negative number

def positive_or_negative(a):
    if a > 0:
        print("Entered number is positive number")
    elif a < 0:
        print("Entered number is negative number")
    else:
        print("Entered number is Zero")


a = int(input("Enter the number: "))
positive_or_negative(a)
```

```python
## two number is equal or not

def equal_or_not(a,b):
    if (a == b):
        print("Both numbers are equal")
    else:
        print("Entered numbers are not equal")

a = int(input("Enter the number: "))
b = int(input("Enter the number: "))

equal_or_not(a,b)

```

```python
## leap year or not

def leapyear_or_not(a):
    if(a%4==0):
        print("Entered year is a leap year")
    else:
        print("Entered year is not a leap year")

a=int(input("enter a year:"))

leapyear_or_not(a)
      
```

```python
## eligible for vote

def eligible_for_vote(age):
    if(age>=18) and (age<100):
         print("you are Eligible for voting")
    elif(age<0) or (age>100):
         print("Enter a valid age")
    else:
         print("you are not eligible for voting")

age=int(input("enter your age"))

eligible_for_vote(age)
```

```python
## height prediction

def height_prediction(ht):
   
   if(ht>=150) and (ht<=200):
      print("you are tall")
   elif(ht>=100) and (ht<150):
      print("you are medium height")
   else:
      print("you are short")

ht=float(input("Enter your height"))

height_prediction(ht)
```

```python
## temperature check

def check_temperature (temp):
    if temp > 30:
        print ("hot")
    if temp>=20 and temp <30:
        print("normal")
    if temp<20:
        print("cold")

temp=float(input("Enter the temperature"))

check_temperature(temp)
    
    
```
