```python
C=[2,7,9,12,15,18,25]

left=0
right=len(C)-1
while (left<right):
    Sum=C[left]+C[right]
    print(f"left :{left} Right{right} Sum={Sum} ")
    if Sum == 30:
       print( C[left],C[right])
       break

    elif Sum <30:
       left+=1

    elif Sum>30:
        right-=1




```

```python
## sliding window techniques

arr = [2, 1, 5, 1, 3, 2]
k = 3

max_sum = 0

for i in range(len(arr) - k + 1):
    window_sum = 0

    for j in range(i, i + k):
        window_sum = window_sum + arr[j]

    if window_sum > max_sum:
        max_sum = window_sum

print(max_sum)
```
