```python
## calculate the gross pay

basic = float(input("Enter Basic Pay:"))
da_percent = float(input("Enter DA%:"))
ta_percent= float(input("Enter TA%:"))

da = basic*da_percent/ 100
ta = basic *ta_percent/100
pf = basic*8/100

gross = (basic+da+ta)-pf

print("gross pay is:", gross)
```

```python
## calculate simple intrest

p = int(input("Enter the principle amount:"))
n = int(input("Enter the time of intrest:"))
r = int(input("Enter the rate of intrest:"))

print ("your simple intrest is:" , (p*n*r)/100)
```

```python
## calculate compound intrest

import math
p = int(input("Enter the principle amount:"))
n = int(input("Enter the n:"))
r = int(input("Enter the rate of intrest:"))
t = int(input("Enter the time in years:"))

print ("your compound intrest is:" , p*math.pow((1+r/n),n*t))
```
