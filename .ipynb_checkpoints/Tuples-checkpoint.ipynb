```python
## creating tuple with value

t=(1,2,3,4)
print(t)
```

```python
## creating tuple with multiple data type

t1=('hello',78,False,7.0)
print(t1)
```

```python
list(t1)
```

```python
## accessing tuple elements
t1[1]
```

```python
## immutable
t1[0]='hi'
```

```python
## slicing tuple
t1[1:]
```

```python
## Tuple methods
data=(1,'hello',2,3,4,5,5,'hi','world',8,9,'world')
data.count('world')
```

```python
data.index('hi')
```

```python
## tuple operations +,*

x=(1,2,3)
y=(4,5,6)
print(x+y)
```

```python
print(x*4)
```

```python
## packing
marks=(100,90,80)
marks
```

```python
## unpacking
science,social,english=marks
print(science,social,english)
```

```python
science=99
marks=science,social,english
print(marks)
```

```python
numbers=(1,2,3,4,5,6)
a,b,c=numbers
```

```python
## unpacking with *
a,b,*c= numbers
print(a)
print(b)
print(c)
```

```python
## list to tuple

num=tuple([1,2,3,4,5,6])
num
```

```python
## tuple to list

num1 = list(('a','b','s','f','r','t'))
num1
```

```python
## nested tuple
ts=((1,2,3),(4,5,6))
print(ts[1][1])
```

```python
print(ts[0][2])
```

```python
for i in ts:
    print (i)
```

```python
## iterating over nested tuples

for item in ts:
    for data in item:
        print(data)
```
