```python
import datetime

dt = datetime.datetime.now()
print(dt)              
print(dt.year)         
print(dt.hour) 
print(dt.minute)
print(dt.strftime('%d-%m-%Y %H:%M')) 


dt2 = datetime.datetime.strptime('10-4-2025 14:30', '%m-%d-%Y %H:%M')
print(dt2)
```

```python
from datetime import datetime,timedelta ##diff of two time or date

today =datetime.today()
print(today)
##Add 10 days
future = today +timedelta(days=20)
print("future date:", future)

## subtract 5 hours

past = today-timedelta(hours=5)
print(past)
```
