```python
## Multiplication tables for 1, 2, 3, 4 & 5 upto 20 rows in each.
table = 1

while table <= 5:
    print("Table", table)

    i = 1
    while i <= 20:
        print(i, "*", table, "=", i * table)
        i += 1

    print()   
    table += 1
```

```python
import random

print("3")
print("5")
print("7")

rounds = int(input("Enter number of rounds: "))

user_score = 0
ai_score = 0

for i in range(1, rounds + 1):

    user_choice = int(input("Enter your choice (1-Rock, 2-Paper, 3-Scissors): "))

    comp = random.randint(1, 3)
    print("AI choice:", comp)

    if comp == 1 and user_choice == 2:
        print("You win")
        user_score = user_score + 1

    elif comp == 2 and user_choice == 3:
        print("You win")
        user_score = user_score + 1

    elif comp == 3 and user_choice == 1:
        print("You win")
        user_score = user_score + 1

    elif comp == user_choice:
        print("Match draw")

    else:
        print("You lose")
        ai_score = ai_score + 1

    print("User score:", user_score)
    print("AI score:", ai_score)

print("Final User Score:", user_score)
print("Final AI Score:", ai_score)

if user_score > ai_score:
    print("You Win")
elif ai_score > user_score:
    print("AI Wins")
else:
    print("Match Draw")
```
