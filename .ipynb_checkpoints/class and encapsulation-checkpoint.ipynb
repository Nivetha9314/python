```python
## adding a method to calculate grades

class student:
    def __init__(self,name,roll_no,marks):
        self.name = name
        self.roll_no =roll_no
        self.marks=marks

    def display(self):
        print(f'student: {self.name}, Roll No: {self.roll_no}, Marks: {self.marks}')

    def cal_grade(self):
        avg=sum(self.marks)/len(self.marks)
        return 'A' if avg >90 else 'B' if avg >=80 else 'C'

    
        
```

```python
student2=student("AAdi",35,[100,90,80])
print(student2.cal_grade())
student2.display()
```

```python
## Access specifier 

class car:
    def __init__(self,model,year):
        self.model=model
        self.year=year
    def display(self):
        print(self.model + " " + self.year)
    
```

```python
c1= car("Toyota","2014")
c1.display()
```

```python
c1.year="2026"
c1.display()
```

```python
c2=car("BMW","2025")
c2.display()
```

```python
class car:
    def __init__(self,model,year):
        self.__model=model
        self.__year=year
        self.price=100000
    def __display(self):
        print(self.__model + " " + self.__year)
    def show(self):
        self.__display()

        
```

```python
c3=car("Benz","2016")
```

```python
c3.show()
```

```python
print(c3.price)
```

```python
## Vehicle 
from datetime import datetime
class vehicle:
    def __init__(self):
        self.__date_time = datetime.now()
        self.__amount=0
        self.__vehicle_no=""
        self.__fuel_name=""
        self.__price_of_air=0

    def getdate_time(self):
        return self.__date_time
        
    def getamount(self):
        return self.__amount

    def setamount(self, amt):
        if amt < 50:
            raise ValueError("Amount should be minimum 50")
        else:
            
            self.__amount = amt

    def getveh_no(self):
        return self.__vehicle_no

    def setveh_no(self,veh):
        self.__vehicle_no=veh

    def getfuel_name(self):
        return self.__fuel_name

    def setfuel_name(self, fu_name):
        self.__fuel_name = fu_name

    def get_price_of_air(self):
        return self.__price_of_air

    def setprice_of_air(self,poa):
        self.__price_of_air=poa

    

```

```python
v1 = vehicle()

print(v1.getdate_time())
```

```python
v1=vehicle()
v1.setamount(100)
v1.getamount()
```

```python
v1.setveh_no("TN45 8056")
v1.getveh_no()
```

```python

v1.setfuel_name("Diesel")
v1.getfuel_name()
```

```python
v1.setprice_of_air(10)
v1.get_price_of_air()
```
