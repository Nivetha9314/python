### Inheritance

```python
class student:
    def __init__(self,name,roll_no):
        self.name=name
        self.roll_no=roll_no

    def _display(self):
        print(f'student: {self.name} , roll_no :{self.roll_no}')

class Graduatestudent(student):
    def __init__(self,name,roll_no,thesis):
        super().__init__(name,roll_no)
        self.thesis=thesis

    def display_thesis(self):
        print(f'thesis name:{self.thesis}')
        
```

```python
gs1=Graduatestudent('shankar',1003,'Embedded system')
print(gs1.display_thesis())
print(gs1._display())
```

```python
class Hostelstudent(student):
    def __init__(self,name,roll_no,hostel_fees):
        super().__init__(name,roll_no)
        self.hostel_fees=hostel_fees

    def display_hostel(self):
        print(f'hostel_fees: {self.hostel_fees}')
        
```

```python
hs1=Hostelstudent('mahi',1204,10000)
hs1.display_hostel()
hs1._display()
```

### Polymorphism

```python
class student():
    def __init__(self,name,rollno):
        self.name=name
        self.rollno=rollno

    def display(self):
        print(f' studentname: {self.name}, studentrollno: {self.rollno}')

class gradstudent(student):
    def __init__(self,name,rollno,thesis):
        super().__init__(name,rollno)
        self.thesis=thesis

    def display(self):
        print(f' hostelstudname: {self.name}, hostelstudrollno: {self.rollno}, thesis: {self.thesis}')
    
```

```python
s1=student("madhu",45)
s2=gradstudent("suganya",30,"Data structure")
```

```python
s1.display()
```

```python
s2.display()
```

```python
ss=[s1,s2]
print(ss)
```

```python
for i in ss:
    print(i.display())
```

### Abstraction

```python
from abc import ABC, abstractmethod

class student(ABC):
    @abstractmethod
    def study(self):
        pass


class undergraduate(student):
    def __init__(self,name):
        self.name=name

    def study(self):
        print(f' {self.name} is studying undergraduate')
```

```python
us1=undergraduate("Ram")
us1.study()
```

```python
from abc import ABC,abstractmethod

class Employee(ABC):
    def __init__(self,name,emp_id,address,mobno,salary):
        self.name=name
        self.emp_id=emp_id
        self.address=address
        self.mobno=mobno
        self._salary=salary

    def getsal(self):
        return(self._salary)

    def setsal(self,val):
        self._salary=val

    @abstractmethod
    def calsalary(self):
        pass

class manager(Employee):
    def __init__(self,name,emp_id,address,mobno,salary):
        super().__init__(name,emp_id,address,mobno,salary)

    def calsalary(self):
        vp=5000
        return self.getsal()+vp
        
    def display(self):
        print(f' managername: {self.name}, managerempid: {self.emp_id}, manageradd: {self.address}, managermob: {self.mobno}, managersal: {self.getsal()}')
```

```python
m1=manager("Nivetha", 1234, "chennai",99989989,80000)
m1.display()
```
